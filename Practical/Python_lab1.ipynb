```python
# --- ARTHEMATIC OPERATORS ---
a = 7
b = 2
print("a + b =",a + b)
print("a - b =",a - b)
print("a * b =",a * b)
print("a / b =",a / b)
print("a ** 2 =",a ** 2)

# --- COMPARISION OPERATORS ---
cgpa = 8.4
print("cgpa == 8.4 :", cgpa == 8.4)
print("cgpa > 8    :", cgpa > 8)
print("cgpa < 6    :", cgpa < 6)

# --- LOGICAL OPERATIONS: and / or / not ---
attendance = 85
placed = (cgpa > 8) and (attendance > 80)
print("placed :",placed)
```

```python
# --- IF / ELIF / ELSE ---
cgpa = 8.2
if cgpa >= 9:
  tier = "High"
elif cgpa >= 7:
  tier = "Mid"
else:
    tier = "Low"
print(f"CGPA {cgpa} falls int tier: {tier}")
missing_count = 0
if missing_count == 0:
  print("No missing values - safe to proceed.")
else:
  print("Missing values found - impute before modelling.")
```
